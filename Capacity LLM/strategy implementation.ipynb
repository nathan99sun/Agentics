```python
import pandas as pd
import numpy as np

# 1. Load Data and Calculate Forward Returns
green_cleaned = pd.read_csv('../green cleaned.csv', dtype={'ncusip': 'string'})
green_cleaned['ret_fwd_1'] = green_cleaned.groupby('permno')['ret_excess'].shift(-1)

variables = ['datadate', 'permno', 'ticker', 'comnam', 'bm', 'mom12m', 'mve', 'ret_fwd_1']
```

```python
start_date = '2021-09-30'

# Filter for the target date
analysis_df = green_cleaned[green_cleaned['datadate'] == start_date][variables].copy()

# 2. Split into Batches (Matching the Agent's Data)
# random_state=42 ensures this split is identical to the one used to create the DBs
df_shuffled = analysis_df.sample(frac=1, random_state=42).reset_index(drop=True)
mid_point = len(df_shuffled) // 2

batch_1 = df_shuffled.iloc[:mid_point].copy()
batch_2 = df_shuffled.iloc[mid_point:].copy()

# =========================================================
# 3. Define and Apply Rules for BATCH 1
# =========================================================
# Insert the rules generated by the LLM for Batch 1 here

# BUY RULE:
# if bm > 1.25 AND mom12m > 1.35:
#     signal = BUY

# SELL RULE:
# if bm < -1.0 OR mom12m < -1.0:
#     signal = SELL

conditions_b1 = [
    # BUY Rule Batch 1
    ( (batch_1['bm'] > 1.25) & (batch_1['mom12m'] > 1.35) ),
    # SELL Rule Batch 1
    ( (batch_1['bm'] < -1.0) | (batch_1['mom12m'] < -1.0) )
]

batch_1['signal'] = np.select(conditions_b1, ['BUY', 'SELL'], default='HOLD')

buys_b1 = batch_1[batch_1['signal'] == 'BUY'][['datadate', 'permno', 'ticker', 'comnam', 'ret_fwd_1']]
sells_b1 = batch_1[batch_1['signal'] == 'SELL'][['datadate', 'permno', 'ticker', 'comnam', 'ret_fwd_1']]

# =========================================================
# 4. Define and Apply Rules for BATCH 2
# =========================================================
# Insert the rules generated by the LLM for Batch 2 here

# BUY RULE:
# if bm > 1.25 AND mom12m > 1.0:
#     signal = BUY

# SELL RULE:
# if bm < -1.0 AND mom12m < -1.0:
#     signal = SELL

conditions_b2 = [
    # BUY Rule Batch 2 (Example rules)
    (batch_2['bm'] > 1.25) & (batch_2['mom12m'] > 1.0),
    # SELL Rule Batch 2 (Example rules)
    (batch_2['bm'] < -1.0) & (batch_2['mom12m'] < -1.0)
]

batch_2['signal'] = np.select(conditions_b2, ['BUY', 'SELL'], default='HOLD')

buys_b2 = batch_2[batch_2['signal'] == 'BUY'][['datadate', 'permno', 'ticker', 'comnam', 'ret_fwd_1']]
sells_b2 = batch_2[batch_2['signal'] == 'SELL'][['datadate', 'permno', 'ticker', 'comnam', 'ret_fwd_1']]

# =========================================================
# 5. Combine, Output, and Save Results
# =========================================================
# Combine the buys and sells from both batches
all_buys = pd.concat([buys_b1, buys_b2], ignore_index=True)
all_sells = pd.concat([sells_b1, sells_b2], ignore_index=True)

print(f"\n{'='*40}")
print(f"COMBINED RESULTS ({len(batch_1) + len(batch_2)} total firms analyzed)")
print(f"{'='*40}")

print(f"\n--- ALL BUYS ({len(all_buys)} firms) ---")
with pd.option_context('display.max_rows', None):
    print(all_buys)

print(f"\n--- ALL SELLS ({len(all_sells)} firms) ---")
print(all_sells)

# Save to single CSVs
all_buys.to_csv('buys_2021.csv', index=False)
all_sells.to_csv('sells_2021.csv', index=False)

print("\n✅ Successfully combined and saved 'buys_2021.csv' and 'sells_2021.csv'")
```
