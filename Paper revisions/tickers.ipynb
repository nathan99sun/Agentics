```python
import pandas as pd

monthly_df = pd.read_csv('../green cleaned.csv', dtype={'ncusip': 'string'})
daily_df = pd.read_csv('filtered_daily_data.csv')

daily_df['date'] = pd.to_datetime(daily_df['date'])
monthly_df['datadate'] = pd.to_datetime(monthly_df['datadate'])
daily_df['year_month'] = daily_df['date'].dt.to_period('M')
monthly_df['year_month'] = monthly_df['datadate'].dt.to_period('M')

daily_df = daily_df.merge(
    monthly_df[['permno', 'year_month', 'ticker']].drop_duplicates(['permno', 'year_month']),
    on=['permno', 'year_month'],
    how='left'
).drop(columns=['year_month'])
```

```python
# Filter using a boolean mask
start_date = '2023-01-01'
end_date = '2024-04-30'

filtered_df = daily_df[(daily_df['date'] >= start_date) & (daily_df['date'] <= end_date)]
```

```python
filtered_df.to_csv('filtered_daily_data_w_ticker.csv', index=False)
```
