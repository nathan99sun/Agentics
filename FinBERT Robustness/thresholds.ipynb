```python
import pandas as pd
import numpy as np

# 1. Load the CSV file
df = pd.read_csv('../examples/monthly_signals_decay.csv')

# 2. Define the new threshold conditions for the weighted_sentiment_score
conditions = [
    df['weighted_sentiment_score'] > 0.15,
    df['weighted_sentiment_score'] < -0.15
]

# 3. Define the corresponding signals
choices = ['buy', 'sell']

# 4. Apply the conditions, using 'hold' as the default for anything in between
df['signal'] = np.select(conditions, choices, default='hold')

# 5. Save the updated data back to a new CSV file
df.to_csv('monthly_signals_decay_0.15.csv', index=False)
```
